```python
%sql
-- 5. Oops: someone deletes all the credits
DELETE FROM workspace.finance.bank_b_clean WHERE direction = 'CREDIT';

SELECT direction, count(*) FROM workspace.finance.bank_b_clean GROUP BY direction;
```

```python
%sql
-- 6. History now shows a DELETE as a new version
DESCRIBE HISTORY workspace.finance.bank_b_clean;
```

```python
%sql
-- 7. Time travel: look at the table BEFORE the mistake
SELECT direction, count(*)
FROM workspace.finance.bank_b_clean VERSION AS OF 0
GROUP BY direction;
```

```python
%sql
-- 8. Undo the mistake
RESTORE TABLE workspace.finance.bank_b_clean TO VERSION AS OF 0;

SELECT direction, count(*) FROM workspace.finance.bank_b_clean GROUP BY direction;
```

```python
%sql
-- 9. The restore itself is recorded too
DESCRIBE HISTORY workspace.finance.bank_b_clean;
```
